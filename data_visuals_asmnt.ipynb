```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Q 1

data = np.random.normal(loc=70, scale=10, size=200)

plt.hist(data, edgecolor ="black", color = "lightblue")

plt.xlabel("SCORE")
plt.ylabel("NO. of students")
plt.grid()

plt.axvline(
    x = np.mean(data),
    color = "red",
    label = "MEAN VALUE",
    linestyle = "--"
)
plt.legend()
```

```python
# Q2
penguins = sns.load_dataset("penguins")

sns.lineplot(
    data = penguins,
    x = "flipper_length_mm",
    y = "body_mass_g",
    hue = "species",
    errorbar = None,
    marker = "o"
)
plt.tight_layout()
```

```python
# Q 3

tips = sns.load_dataset("tips")

sns.boxplot(
    data = tips,
    x = "day",
    y = "total_bill",
    hue = "sex"
)
plt.tight_layout()

import pandas as pd

tips.groupby("day")["total_bill"].median()
```

```python
# Q4

tips = sns.load_dataset("tips")

sns.scatterplot(
    data = tips,
    x ="total_bill",
    y = "tip",
    hue = "sex",
    style = "smoker")

plt.title("TOTAL BILL VS TIPS ")
plt.xlabel("total bill")
plt.ylabel("bill")
```

```python
# Q5

heatmap_data = tips.pivot_table(
    index = "day",
    columns = "time", 
    values = "total_bill",
    aggfunc= "mean")

sns.heatmap(
    heatmap_data,
    annot=True, 
    fmt=".2f")
    
```
