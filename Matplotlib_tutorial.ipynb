```python
# BAR CHART

oscar_movies = [
    "The Dark Knight", 
    "The Hurt Locker",  
    "The King's Speech", 
    "The Artist",     
    "Argo"
]
oscar_years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M

plt.bar(oscar_years, oscar_revenue, color = "purple")
plt.title("OSCAR MOVIES REVENUE IN EACH YEAR")
plt.xlabel("oscar years")
plt.ylabel("oscar revenue")

for i in range(len(oscar_years)):
    plt.text(oscar_years[i], oscar_revenue[i]+10, str(oscar_revenue[i]), ha = "center")

plt.ylim(0, max(oscar_revenue)+100)  # extend y axis val
```

```python
# COMPARING DATA WITH BAR CHART

oscar_movies = [
    "The Dark Knight", 
    "The Hurt Locker",  
    "The King's Speech", 
    "The Artist",     
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M


non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
years = [2008, 2009, 2010, 2011, 2012]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M


import numpy as np

x = np.arange(len(years))
width = 0.35

plt.bar(x-width/2, oscar_revenue, width, color = "purple", label = "oscar revenue")
plt.bar(x+width/2, non_oscar_revenue, width, color = "blue", label = "non oscar revenue")

plt.title("comparison of oscar vs non oscar movies' revenue")
plt.xlabel("years")
plt.ylabel("revenue in (M$)")
plt.legend()
plt.tight_layout()

plt.xticks(x, years)
```

```python
# SCATTER PLOTS

people = ["Person A", "Person B", "Person C", "Person D", "Person E", 
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

colors = ["red" if x > 135  else "green" for x in bp]


# plt.scatter(age, bp, color = colors, s = bp, marker = "^", alpha = 0.2)
plt.scatter(age, bp, s = bp, marker = "^", cmap = "plasma", c = bp)
plt.title("AGE VS BP")
plt.xlabel("age")
plt.ylabel("blood pressure")

plt.grid()

plt.colorbar(label = "BP")

for i in range(len(people)):
    plt.annotate(people[i], xy = (age[i], bp[i]), xytext = (age[i]+1, bp[i]+1))

plt.xlim(min(age), max(age)+10)
plt.ylim(min(bp), max(bp)+5)
"""
alpha ---> for transparancy which lies from (0 to 1)
s -----> for side of dots (bp) mean for large val large circle and so on
cmap ----> matplotlib has some spectrum of colors like plasma, OrRd etc which gives color according to the val
"""
```

```python
# Pie Charts
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]

explode = [0, 0.2, 0, 0.2, 0]

plt.style.use("default")
plt.pie(amounts, labels =expenses, autopct = "%1.2f%%",
        wedgeprops =
            {"edgecolor": "black",
             "linestyle": "--",
             "linewidth": 1},
       explode = explode,
       startangle = 60,
       shadow = True
       )        
plt.show()
```

```python
# HISTOGRAMS

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

scores = np.random.normal(70, 10, 1000)

bins = [25, 50, 75, 100]

plt.hist(scores, color = "skyblue", edgecolor = "black", bins = bins)

plt.xlabel("NO. of students")
plt.ylabel("scores")

```

```python
# MULTIPLE DATA SETS

legit_transactions = [
    2.99, 5.49, 8.99, 12.50, 14.99, 19.99, 23.45, 29.99, 34.99, 39.50,
    45.00, 49.99, 55.25, 60.00, 75.99, 89.99, 120.50, 150.00, 199.99,
    249.99, 300.75, 450.00, 600.00, 850.00, 1200.00
]

fraud_transactions = [
    50, 100, 150, 200, 300, 500, 500, 750, 1000, 1000,
    1200, 1500, 1500, 2000, 2500, 3000, 3000
]

plt.hist(fraud_transactions, color = "red", alpha = 0.5, bins = 10, label = "FRAUD")
plt.hist(legit_transactions, color = "green", alpha = 0.5, bins = 10, label = "LEGIT")

plt.title('Transaction Amount Distribution: Legitimate vs Fraudulent')
plt.xlabel("AMOUNT IN ($)")
plt.ylabel("FREQUENCY")
plt.legend()

plt.grid()
```

```python
# axv line
marks = [
    25, 38, 42, 55, 61, 67, 72, 78, 83, 88,
    91, 95, 47, 53, 59, 64, 69, 74, 79, 85,
    92, 97, 33, 45, 58, 62, 71, 76, 81, 89
]

# custom bins
bins = [0, 30, 70, 90, 100]
plt.hist(
    marks,
    bins=bins,
    edgecolor="black",
    color="lightblue"
)

# line at passing marks - 33
plt.axvline(
    x=33,
    color="red",
    linestyle="--",
    linewidth=2,
    label="Pass Mark (33)"
)

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Marks of 30 Students")
plt.grid(True)

```

```python
# Stack Plot
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
direct = [50, 60, 70, 80, 90, 100, 110]
organic = [30, 40, 50, 55, 60, 70, 80]
social = [20, 25, 30, 35, 40, 50, 60]

plt.stackplot(days, [direct, organic, social], labels = ["direct", "organic", "social"])
plt.title("MARKETING DATA OVER THE WEEK")
plt.xlabel("DAYS")
plt.ylabel("NO. OF CUSTOMERS")
plt.legend()
```

```python
# SUB PLOTS

x = [1, 2, 3, 4, 5, 6, 7, 8]

y1 = [np.sqrt(i) for i in x]
y2 = [i*2 for i in x]
y3 = [i**2 for i in x]
y4 = [i**3 for i in x]

plt.subplot(2, 2, 1)        # row, col, index
plt.plot(x, y1)
plt.title("SQUARE ROOT")     # SQUARE ROOT 

plt.subplot(2, 2, 2)
plt.plot(x, y2)
plt.title("DOUBLE")     # DOUBLE

plt.subplot(2, 2, 3)
plt.plot(x, y3)
plt.title("DOUBLE")   # DOUBLE

plt.subplot(2, 2, 4)
plt.plot(x, y4)
plt.title("CUBES")   # CUBES

plt.tight_layout()
```

```python
# Modern Matplotlib - oop

fig, axes = plt.subplots(2,2)

axes[0][0].plot(x, y1)
axes[0][0].set_title("Plot 1 - square root")

axes[0][1].plot(x, y2)
axes[0][1].set_title("Plot 2 - DOUBLE")

axes[1][0].plot(x, y3)
axes[1][0].set_title("Plot 3 - square")

axes[1][1].plot(x, y4)
axes[1][1].set_title("Plot 4 - CUBES")

fig.suptitle("Multiple plots")
fig.tight_layout()
```

```python
# Temperature over cities
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]

cities = ["New York", "London", "Delhi", "Tokyo"]

temperatures = [
    [22, 23, 21, 24, 25],   # New York
    [18, 19, 17, 20, 21],   # London
    [30, 32, 31, 33, 34],   # Delhi
    [25, 26, 24, 27, 28]    # Tokyo
]

fig, axes = plt.subplots(2, 2)
city_no = 0
for i in range(2):
    for j in range(2):
        axes[i][j].plot(days, temperatures[city_no], marker = "o", color = "pink")
        axes[i][j].set_title(cities[city_no])
        axes[i][j].grid()
        city_no +=1

fig.suptitle("TEMP IN CITIES OVER THE WEEK")
fig.tight_layout()
```
